```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt

data = pd.read_csv('20260711_Rexburg_Idaho.csv')

y1 = data.iloc[0,5:27].values
y2 = data.iloc[1,5:27].values
y3 = data.iloc[2,5:27].values
y4 = data.iloc[3,5:27].values
y5 = data.iloc[4,5:27].values
y6 = data.iloc[5,5:27].values

x = data.columns[5:27].astype(float).values
# The time labels are local time at Rexburg instead of UTC.
plt.plot(x, y1, '.-', label="0711 - 13:31:45")
plt.plot(x, y2, 'o-', label="0711 - 14:08:53")
plt.plot(x, y3, '*-', label="0711 - 15:07:11")
plt.plot(x, y4, '^-', label="0711 - 15:40:38")
plt.plot(x, y5, 'o-', label="0711 - 15:59:29")
plt.plot(x, y6, '.-', label="0711 - 16:43:27")
plt.xscale('log')
plt.ylabel(r'$\frac{dV(r)}{dln(r)}$ [$\frac{\mu m^3}{\mu m^2}$]',fontsize=14)
plt.xlabel(r'Radius (r) [$\mu m$]')
plt.suptitle('Size Distribution Almucantar Level 1.5 (NASA)')
plt.title('July 11, 2026',fontsize=10)
plt.legend()
plt.show()
```
