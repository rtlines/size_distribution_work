```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt

data = pd.read_csv('20260509_Rexburg_Idaho.csv')

y1 = data.iloc[0,5:27].values
y2 = data.iloc[1,5:27].values
y3 = data.iloc[2,5:27].values
y4 = data.iloc[3,5:27].values
y5 = data.iloc[4,5:27].values

x = data.columns[5:27].astype(float).values

plt.plot(x, y1, 'r.-', label="08:18:17")
plt.plot(x, y2, 'bo-', label="09:16:57")
plt.plot(x, y3, 'y*-', label="09:51:55")
plt.plot(x, y4, 'g^-', label="10:34:51")
plt.plot(x, y5, 'ko-', label="16:34:56")
plt.xscale('log')
plt.ylabel(r'$\frac{dV(r)}{dln(r)}$ [$\frac{\mu m^3}{\mu m^2}$]',fontsize=14)
plt.xlabel(r'Radius (r) [$\mu m$]')
plt.suptitle('Size Distribution Almucantar Level 1.5 (NASA)')
plt.title('May 9, 2026',fontsize=10)
plt.legend()
plt.show()
```
